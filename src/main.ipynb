# Data Cleaning and Preprocessing

```python
# Creating another copy so that we don't change the original dataset
df  = ds.copy()
```

```python
df.head()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.shape
```

```python
df.dropna()
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
df.info()
```

```python
df["sex"]
```

```python
# we need to encode all str , objects to int 
df["sex"] = df["sex"].map({"male":0, "female": 1})
# male is 0 and female is 1
```

```python
df["smoker"] = df["smoker"].map({"no": 0, "yes": 1})
```

```python
df
```

```python
df.rename(columns={
    "sex":"is_female",
    "smoker": "is_smoker"
},inplace=True)
```

```python
df
```

```python
df["region"].value_counts()
```

```python
# we will use one hot encoding
df = pd.get_dummies(df, columns=["region"])
```

```python
df.head()
```

```python
df = df.astype(int)
```

```python

df.head()
```

# Feature Engineering and Extraction

```python
# creating based on bmi like , underweight , normal , over weight obesity 
df["bmi_category"] = pd.cut(
    df["bmi"],
    bins=[0, 18.5, 24.9, 29.9, float('inf')], # flot till infinity
    labels=["under_weight", "normal", "over_weight", "obesity"]
)
```

```python
df.head()
```

```python
# but we need to make it into integers
# also we can change anything more but shouldn't change the charges as we need to train and predict those
df = pd.get_dummies(df, columns=["bmi_category"])
```

```python
df.head()
```

```python
df = df.astype(int)
```

```python
df.head()
```
